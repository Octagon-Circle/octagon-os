```python
# 4. Descargar automáticamente el archivo .iso generado a tu ordenador
import glob
from google.colab import files

isos = glob.glob("*.iso")
if isos:
    print(f"Descargando {isos[0]}...")
    files.download(isos[0])
else:
    print("No se encontró el archivo ISO. Revisa la celda anterior.")
```
